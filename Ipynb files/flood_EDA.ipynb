## **Visualize Numerical Feature Distributions**

**To visualize numerical feature distributions, I will first identify all numerical columns in the DataFrame. Then, for each numerical column, I will generate a histogram to show the distribution and a box plot to identify potential outliers, utilizing matplotlib.pyplot and seaborn for effective visualization.**

**Separate the DataFrame columns into numerical and categorical types.**

```python
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

print("Numerical Columns:", numerical_cols)
print("Categorical Columns:", categorical_cols)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Loop through each numerical column and plot its distribution
for col in numerical_cols:
    plt.figure(figsize=(15, 5))

    # Histogram subplot
    plt.subplot(1, 2, 1)
    sns.histplot(df[col].dropna(), kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    # Box plot subplot
    plt.subplot(1, 2, 2)
    sns.boxplot(y=df[col].dropna())
    plt.title(f'Box Plot of {col}')
    plt.ylabel(col)

    plt.tight_layout()
    plt.show()
```

## Visualize Categorical Feature Distributions

### Subtask:
Generate count plots for each categorical column to display the frequency distribution of their unique values. Each plot will include appropriate labels and legends.


**Reasoning**:
To visualize the distributions of categorical features, I will iterate through each categorical column and generate a count plot for each, setting appropriate titles and labels for readability.



```python
for col in categorical_cols:
    plt.figure(figsize=(12, 6))
    sns.countplot(data=df, x=col, palette='viridis')
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```
